```
import torch, torchaudio as ta
from IPython.display import Audio, display
from chatterbox.mtl_tts import ChatterboxMultilingualTTS

device = 'cuda' if torch.cuda.is_available() else 'cpu'
model = ChatterboxMultilingualTTS.from_pretrained(device=device)
print('Modèle prêt sur', device)
```

## Voix par défaut

```
texte = "Bonjour, bienvenue sur mon portfolio. Je crée des animations et des univers inspirés des animes."
wav = model.generate(texte, language_id='fr', exaggeration=0.5, cfg_weight=0.5)
ta.save('voix_defaut.wav', wav, model.sr)
display(Audio('voix_defaut.wav'))
```

## Cloner une voix
Importez un fichier `ma_voix.wav` (10 secondes, sans bruit de fond) via l'icône dossier à gauche.

```
wav = model.generate(texte, language_id='fr', audio_prompt_path='ma_voix.wav', exaggeration=0.5, cfg_weight=0.5)
ta.save('voix_clonee.wav', wav, model.sr)
display(Audio('voix_clonee.wav'))
```
